## 1) Типы данных: числа, строки и коллекции

Python — язык с **динамической типизацией**: тип хранится у значения, а не “жёстко” у переменной.

### Числа
- **`int`**: целые числа (в Python произвольной длины).
- **`float`**: числа с плавающей точкой.
- **`complex`**: комплексные числа (например, `1+2j`).

Нюанс: у `float` есть погрешности представления (это нормально для бинарной арифметики).



```python
# Примеры: числа, строки, bool, None, bytes

# Числа
x = 10          # int
y = 3.5         # float
z = 1 + 2j      # complex
print('x: ', type(x), x)
print('y: ', type(y), y)
print('z: ', type(z), z)
# Погрешности float
print(0.1 + 0.2)           # не всегда ровно 0.3
print(round(0.1 + 0.2, 2)) # иногда округляют для вывода
```

### Строки (`str`)
- **Что это**: последовательность символов Unicode.
- **Особенности**:
  - **Неизменяемая (immutable)**: нельзя менять символ “на месте”.
  - Поддерживает **индексацию** и **срезы**.
  - Много методов: `.lower()`, `.upper()`, `.strip()`, `.split()`, `.replace()`.
  - Форматирование: **f-строки** (`f"...{x}..."`).

```python
# Строки (str)
s = "Hello"
print("s=", s, "len=", len(s))
print("s[1]=", s[1])
print("s[1:4]=", s[1:4])

# Строки неизменяемы:
# s[0] = "h"  # TypeError


# Полезные методы
raw = "   PyThOn, python  "
print(raw.strip())
print(raw.lower())
print(raw.replace("python", "PY"))
print(raw.split(","))

# f-строки
name = "Alice"
age = 25
print(f"{name} is {age} years old")
```

### Логический тип (`bool`)
- Значения: `True` и `False`.
- Получается из сравнений (`a < b`) и логических операций (`and/or/not`).

### `None`
- Специальное значение “ничего/нет значения”.
- Часто используется как “значение по умолчанию” или “не найдено”.

### Байты (`bytes`) и `bytearray`
- **`bytes`**: неизменяемая последовательность байтов (полезно для файлов/сети).
- **`bytearray`**: изменяемая версия.


```python
# bool и логика
print(5 > 3, 5 == 3, 5 != 3)
print(True and False, True or False, not True)

# None
value = None
print("value is None?", value is None)

# bytes / str: кодировки
text = "Hello"
b = text.encode("utf-8")
print("bytes:", b)
print("decoded:", b.decode("ASCII"))

# Преобразования типов
print(int,("123"))
print(float("3.14"))
print(str(10))
print(bool(""), bool("0"), bool(0), bool(1))
```

### Коллекции (основные)

#### Список (`list`)
- **Что это**: упорядоченная коллекция элементов.
- **Особенности**:
  - **Изменяемый (mutable)**: можно менять элементы, добавлять/удалять.
  - Может хранить **разные типы** внутри.
  - Поддерживает **индексацию** и **срезы**.

#### Кортеж (`tuple`)
- **Что это**: упорядоченная коллекция, похожая на список.
- **Особенности**:
  - **Неизменяемый (immutable)**: нельзя менять элементы после создания.
  - Часто используется как “запись” (record) или ключ словаря.

#### Множество (`set`)
- **Что это**: неупорядоченная коллекция **уникальных** элементов.
- **Особенности**:
  - Быстрые операции: **проверка принадлежности**, **пересечения**, **объединения**.
  - Элементы должны быть **хешируемыми** (обычно неизменяемые типы).

#### Словарь (`dict`)
- **Что это**: отображение **ключ → значение**.
- **Особенности**:
  - Ключи — **уникальные** и **хешируемые**.
  - Удобен для хранения структурированных данных.

### Преобразование типов (casting)
- `int("123")`, `float("3.14")`, `str(10)`, `bool(x)`.
- Важно: `bool("0")` будет `True`, потому что непустая строка — truthy.

### “Массив”
В Python под словом “массив” могут иметь в виду разные вещи:
- **`list`** (часто в быту так и называют)
- **`array.array`** из стандартной библиотеки: компактный массив **однотипных** чисел
- **`numpy.ndarray`** (если установлен NumPy): быстрые векторные операции

```python
dict = {'job': 'analytic', 10: {'python': 'expert'}}
dict[10]['python']
```

```python
# Примеры: list / tuple / set / dict

lst = [1, 2, 3]
lst.append(4)
lst[0] = 10
print("list:", lst, "len=", len(lst))
print("срез lst[1:3] =", lst[1:3])

tpl = ("Alice", 25)
print("tuple:", tpl)
# tpl[0] = "Bob"  # Ошибка: кортеж неизменяем

s = {1, 2, 2, 3}
print("set (уникальные):", s)
print("2 in set?", 2 in s)
print("операции множеств:", {1,2,3} | {3,4}, {1,2,3} & {3,4}, {1,2,3} - {3,4})

d = {"name": "Alice", "age": 25}
d["age"] += 1
print("dict:", d)
print("keys:", list(d.keys()))
print("items:", list(d.items()))

```

```python
# Массивы: стандартный array.array и NumPy (если доступен)
from array import array
import numpy as np

arr = array("i", [1, 2, 3])  # "i" = signed int
arr.append(4)
print("array.array:", arr, "typecode=", arr.typecode)

# NumPy массив (ndarray): быстрые поэлементные операции

a = np.array([1, 2, 3])
b = a * 10 + 1
print("numpy a:", a, "dtype=", a.dtype)
print("numpy b = a*10+1:", b)
```

## 2) Циклы: `for`, `while`

### `for`
- Используется для **перебора** элементов коллекции (списка, строки, словаря, множества) или последовательности из `range()`.
- Частые помощники:
  - `range(n)` — числа 0..n-1
  - `enumerate(iterable)` — пары (индекс, элемент)
  - `break` — досрочно выйти из цикла
  - `continue` — перейти к следующей итерации
  - `pass` - ничего не делать

### `while`
- Цикл выполняется, пока условие истинно.
- Важно следить, чтобы условие **когда-то стало ложным** (иначе будет бесконечный цикл).

### `else` у циклов (редко, но полезно)
- Блок `else` у `for/while` выполнится, **если цикл завершился без `break`**.



```python
if 1!=1:
    print('rgre')
elif 2!=2:
    print('erg')
elif 3!=3:
    print('xyz')
else:
    print('')
```

```python
for x in [1, 2, 3, 4, 5, 6, 7, 8]:
    if x == 3:
        continue
    if x == 5:
        break
    print("x=", x)
```

```python
for y in range(3):
    print(y)
```

## 3) Условия: `if` / `elif` / `else`

- `if` проверяет условие (выражение, которое даёт `True/False`).
- `elif` — “иначе если” (может быть много).
- `else` — “иначе” (выполняется, если ни одно условие выше не сработало).

### Частые нюансы
- **Сравнения**: `==`, `!=`, `<`, `<=`, `>`, `>=`
- **Логика**: `and`, `or`, `not`
- **Принадлежность**: `x in collection`
- **Истинность (truthy/falsy)**:
  - `False`, `0`, `0.0`, `""`, `[]`, `{}`, `set()`, `None` считаются “ложными”
  - остальное обычно “истинно”
- **Тернарный оператор**: `a if condition else b`



```python
# Примеры if/elif/else

x = 7
if x < 0:
    print("x отрицательный")
elif x == 0:
    print("x равен нулю")
else:
    print("x положительный")

# and/or/not
age = 20
has_ticket = True
if age >= 18 and has_ticket:
    print("можно проходить")

# in
letters = {"a", "b", "c"}
print("a in letters?", "a" in letters)

# truthy/falsy
values = [0, 1, "", "hi", [], [1], None]
for v in values:
    if v:
        print(repr(v), "=> truthy")
    else:
        print(repr(v), "=> falsy")

# Тернарный оператор
score = 83
grade = "pass" if score >= 60 else "fail"
print("grade=", grade)

```
