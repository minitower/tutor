```python
class NumArray:
    def __init__(self, nums):
        self.cumulative_nums = []
        for i in nums:
            if len(self.cumulative_nums)==0:
                self.cumulative_nums.append(i)
            else:
                self.cumulative_nums.append(i+self.cumulative_nums[-1])

    def sumRange(self, left: int, right: int) -> int:
        if left == 0:
            return self.cumulative_nums[right]
        else:
            return self.cumulative_nums[right] - self.cumulative_nums[left - 1]

NumArray([-2, 0, 3, -5, 2, -1]).sumRange(0, 2)
```

```python
# Создание и изменение столбцов

# Цена в тысячах долларов
df['Price_K_USD'] = df['Price_USD'] / 1000

# Соотношение мощность/объём двигателя (л.с. на 1 куб. см)
df['HP_per_CC'] = df['Horsepower'] / df['Engine_CC']

df[['Car_ID', 'Price_USD', 'Price_K_USD', 'HP_per_CC']].head()
```

```python
# Агрегирование и сводные таблицы

brand_stats = df.groupby('Brand').agg({
    'Price_USD': 'mean',
    'Horsepower': 'mean',
    'Engine_CC': 'mean'
}).round(1)

brand_stats.head()
```

```python
# Сводная таблица (pivot_table)

# Средняя цена по типу кузова и типу топлива
pivot_price = pd.pivot_table(
    df,
    values='Price_USD',
    index='Body_Type',
    columns='Fuel_Type',
    aggfunc='mean'
).round(0)

pivot_price
```

```python
# Объединение DataFrame (merge, concat)

# Пример: создадим искусственную таблицу с налоговыми ставками по странам
country_tax = pd.DataFrame({
    'Manufacturing_Country': df['Manufacturing_Country'].unique(),
    'Tax_Rate': [0.05, 0.08, 0.1, 0.12, 0.15, 0.2][:len(df['Manufacturing_Country'].unique())]
})

# Объединим с основным датафреймом по стране производства
merged_df = df.merge(country_tax, on='Manufacturing_Country', how='left')
merged_df[['Car_ID', 'Manufacturing_Country', 'Tax_Rate']].head()
```

```python
country_tax = pd.DataFrame({
    'Manufacturing_Country_': df['Manufacturing_Country'].unique(),
    'Tax_Rate': [0.05, 0.08, 0.1, 0.12, 0.15, 0.2][:len(df['Manufacturing_Country'].unique())]
})

df.join(country_tax.set_index('Manufacturing_Country_'), on='Manufacturing_Country', how='left')
```

```python
# Построение DataFrame для линейного графика
mean_price = df[['Manufacture_Year', 'Price_USD']].groupby('Manufacture_Year').mean()
mean_price.sort_index()
```

```python
fig_3 = plt.figure(figsize=(10,6))
plt.plot(mean_price.index.astype(str), mean_price.values, marker='o')
plt.title('Тенденция цен на машины с разным годом выпуска')
plt.xlabel('Год выпуска')
plt.xticks(rotation=45)
plt.ylabel('Цена')
plt.show()
```

```python
# Объединение трёх графиков в одном окне (3 подграфика)

fig, axes = plt.subplots(2, 2, figsize=(12, 6))

# 1. Гистограмма цен
axes[0, 0].hist(df['Price_USD'], bins=30, edgecolor='black')
axes[0, 0].set_title('Распределение цен автомобилей')
axes[0, 0].set_xlabel('Цена, USD')
axes[0, 0].set_ylabel('Количество')

# 2. Столбчатая диаграмма мощности по типу кузова
axes[1, 0].bar(x=grouped_power.index,
            height=grouped_power['Horsepower'],
            edgecolor='black',
            color='green')
axes[1, 0].set_title('Распределение мощности автомобилей по типу кузова')
axes[1, 0].set_xlabel('Тип кузова')
axes[1, 0].set_ylabel('Мощность, л.с.')

# 3. Линейный график средней цены по году выпуска
axes[0, 1].plot(mean_price.index.astype(str), mean_price.values, marker='o')
axes[0, 1].set_title('Тенденция цен на машины с разным годом выпуска')
axes[0, 1].set_xlabel('Год выпуска')
axes[0, 1].set_ylabel('Средняя цена, USD')
axes[0, 1].tick_params(axis='x', rotation=45)

axes[1, 1].plot(mean_price.index.astype(str), mean_price.values, marker='o')
axes[1, 1].set_title('Тенденция цен на машины с разным годом выпуска')
axes[1, 1].set_xlabel('Год выпуска')
axes[1, 1].set_ylabel('Средняя цена, USD')
axes[1, 1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```
