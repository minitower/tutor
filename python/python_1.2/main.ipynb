```python
class Bicycle:
    def __init__(self, color:str, price:tuple):
        self.color = color
        self.price = price
        
    def get_price(self):
        return self.price+10

    def get_color(self):
        return self.color
    
    def __private_method(self):
        return "This is a private method"
```

```python
class Car(Bicycle):
    def __init__(self, color:str, price:tuple):
        super().__init__(color, price)
        self.color = color
        self.price = price
        
    def get_price(self):
        return 4
```

```python
Car(color="blue", price=(50, 60)).get_price()
```
