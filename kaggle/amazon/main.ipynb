```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('./Amazon.csv')
```

```python
df
```

```python
from scipy.stats import spearmanr

spearman_corr = spearmanr(df['Quantity'], df['TotalAmount'])
spearman_corr
```

```python
from scipy.stats import pearsonr

pearson_corr = pearsonr(df['Quantity'], df['TotalAmount'])
pearson_corr
```

```python
import seaborn as sns

sns.heatmap(df[['Quantity', 'TotalAmount', 'Discount', 'Tax', 'ShippingCost']].corr(), annot=True, cmap='coolwarm')
```

```python
df['true_CustomerID'] = df['CustomerID'] + '_'+ df['CustomerName']
```

```python
df_success = df.loc[df['OrderStatus'] == 'Delivered']
```

```python
df_success[['TotalAmount', 'true_CustomerID']].groupby('true_CustomerID').sum().sort_values('TotalAmount', ascending=False).head(10)
```

```python
df_success[['Country', 'TotalAmount']].groupby('Country').sum().sort_values('TotalAmount', ascending=False)
```

```python
piv_table = df_success[['Country', 'TotalAmount', 'PaymentMethod']].pivot_table(index='Country', columns='PaymentMethod', values='TotalAmount', aggfunc='sum').fillna(0)
```

```python
import seaborn as sns

sns.heatmap(piv_table, annot=True, fmt='.0f', cmap='Blues')
```

```python
df.loc[(df['OrderStatus'] == 'Cancelled') | (df['OrderStatus'] == 'Returned')][['true_CustomerID', 'OrderStatus']].groupby('true_CustomerID').count().sort_values('OrderStatus', ascending=False).head(10)
```

```python
cat_arr = df['Category'].unique()

for i in cat_arr:
    plt.figure(figsize=(10, 6))
    plt.hist(df.loc[df['Category'] == i]['UnitPrice'], bins=20, alpha=0.7)
    plt.title(f'Unit Price Distribution for Category: {i}')
    plt.xlabel('Unit Price')
    plt.ylabel('Frequency')
    plt.show()
```
