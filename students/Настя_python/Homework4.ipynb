# Задание 1

```python
def arcsin(x):
    true_val = math.asin(x)
    k = 0
    term = x # член суммы при k=0
    err_arr = []
    k_arr = []
    result = term
    err_arr.append(abs(true_val-result))
    k_arr.append(k)
    while True:
        k+=1
        term = (math.factorial(2*k)/(2**(2*k)*(math.factorial(k)**2)))*(x**(2*k+1)/(2*k+1))
        if abs(term) < 1e-10:
            #term_arr.append(term)
            break
        result += term
        k_arr.append(k)
        err_arr.append(abs(true_val-result))
    return err_arr, k_arr
```

```python
y, x = arcsin(0.5)
plt.figure(figsize=(10, 6))
plt.plot(x, y)
plt.title('График абсолютной погрешности arcsin(0.5) до 10*e^-10')
plt.xlabel('Количество итераций')
plt.ylabel('Погрешность')
```

# Задание 2

```python
def y1(x):
    return 1.2*x**6 - 14*x**5 +5.2*x**4-8.3*x**3+x+2

def y2(x):
    return 1/x**3

arr = np.linspace(0, 10, 100_000)
y1_res = []

for x in arr:
    y1_res.append(y1(x))
```

```python
y1_val = np.max(np.abs(y1_res))

x = arr[y1_res.index(np.max(y1_res))]
```

```python
M = y1_val

def f(L):
    return abs(y2(L))-M

a = 1e10


b = 1e-10


L = bisect(f=f, a=a, b=b)

print(L)
print(M)
```

```python
arr_x = np.linspace(L, 10, 100000)

arr_y1 = []
arr_y2 = []

for x in arr_x:
    arr_y1.append(y1(x))
    arr_y2.append(y2(x))    
    
```

```python
plt.figure(figsize=(10, 6))
plt.plot(arr_x, y1_res, label='y1')
plt.plot(arr_x, arr_y2, label='y2')
plt.ylim([-M, M])
plt.legend()
```

```python
def f1(x):
    return np.sin(1000*y1(x)/M)

def f2(x):
    return np.sin(1000*y2(x)/M)

arr_x = np.linspace(0.01, 10, 10000)

arr_f1 = f1(arr_x)
arr_f2 = f2(arr_x)

plt.figure(figsize=(10, 6))
plt.subplot(211)
plt.plot(arr_x, arr_f1, label='f1(x)')
plt.plot(arr_x, arr_f2, label='f2(x)')
plt.xlabel("x")
plt.ylabel("y1, y2")
plt.title('Графики функции f1(x), f2(x) в линейном масштабе')
plt.legend()

arr_x_log = np.linspace(0.01, 10, 1000000)

arr_f1_log = f1(arr_x_log)
arr_f2_log = f2(arr_x_log)

plt.subplot(212)
plt.plot(arr_x_log, arr_f1_log, label='f1(x)')
plt.plot(arr_x_log, arr_f2_log, label='f2(x)')
plt.xlabel("x")
plt.ylabel("y1, y2")
plt.xscale('log')
plt.title('Графики функции f1(x), f2(x) в лог. масштабе')
plt.legend()
```
