```python
import math
```

```python
def arcsin(x):
    k = 0
    term = x # член суммы при k=0
    result = term
    while True:
        k+=1
        term = (math.factorial(2*k)/(2**(2*k)*(math.factorial(k)**2)))*(x**(2*k+1)/(2*k+1))
        if abs(term) < 1e-10:
            break
        result += term
    return result, k
```

```python
res = 0
for k in range(50):
    res+=(math.factorial(2*k)/(2**(2*k)*(math.factorial(k)**2)))*(0.1**(2*k+1)/(2*k+1))
print(res)
```
