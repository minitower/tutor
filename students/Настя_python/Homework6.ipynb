```python
import numpy as np
#import warnings
#warnings.filterwarnings("ignore", category=np.ComplexWarning)
import matplotlib.pyplot as plt

def y1(x):
    return 1.2*x**6 -14*x**5 + 5.2*x**4 -8.3*x**3 + x + 2


coeffs = [1.2, -14, 5.2, -8.3, 0, 1, 2]

roots = np.roots(coeffs)

for x in roots:
    y = y1(x)
    print(f"f({x}) = {y}")
    
x = np.linspace(-1, 12, 1000)

y_arr = y1(x)
roots_visual = np.where(np.iscomplex(roots)==False, roots, np.nan)


plt.figure(figsize=(10, 6))
plt.plot(x, y_arr, label='y1(x)')
plt.scatter(roots_visual.astype(float), [0]*len(roots_visual), color='red', label='Корни')
plt.title('График функции y1(x) и её корни')
plt.axhline(0) 
plt.xlabel('x')
plt.ylabel('y1(x)')
plt.legend()
plt.grid()

```

```python
import numpy as np
def f(x,w=8):
    return w * np.cos(x)+3*x**2 * np.sin(x)

(x_min, x_max) = (-8, 8)

X_MIN_GLOBAL = x_min
X_MAX_GLOBAL = x_max

n_shag=0.1
n_round=1
xres=x_min
tol = 1e-10

while n_shag > tol:
    x_value = np.arange(x_min, x_max, n_shag)
    
    y_value = f(x_value)
    max_value = -np.inf
    for y in y_value:
        if y > max_value:
            max_value = y
            xres = round(x_value[np.where(y_value==max_value)][0], n_round)

    x_min = max(X_MIN_GLOBAL, xres - n_shag)
    x_max = min(X_MAX_GLOBAL, xres + n_shag)
    n_round+=1
    n_shag/=10
    n_shag = round(n_shag, n_round)
    print(xres, max_value, n_shag, x_max)
```
