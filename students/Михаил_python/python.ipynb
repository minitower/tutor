# Задача 1

```python
book_arr_str = 'Война и мир, Преступление и наказание, Маленький принц, Идиот'#input()
len_book_arr = '1225, 671, 96, 641'#input()

book_arr = book_arr_str.split(', ')
try:
    len_book = [int(x) for x in len_book_arr.split(', ')]
except ValueError:
    print('Длина книги не число')

avg_len_book = sum(len_book)/len(len_book)

k = 0
res_arr = []
res_len = []

for lenght in len_book:
    if lenght>=avg_len_book:
        res_arr.append([book_arr[len_book.index(lenght, k)], lenght])
        res_len.append(lenght)
        k+=1

res_len.sort()

for i in res_len:
    for ii in res_arr:
        if i==ii[1]:
            print(ii[0])
```

# Задача 2

```python
needed_book_str = 'Властелин колец; Грозовой перевал; Хоббит'#input() #'Война и мир; Преступление и наказание; Отцы и дети'
lib_has_str = 'Хоббит; Властелин колец; Грозовой перевал'#input()# 'Маленький принц; Война и мир; Мастер и Маграгита; 1984'
vitya_want_str = 'Гордость и предубеждение'#input()#'1984; Идиот'

needed_arr = needed_book_str.split('; ')
lib_has_arr = lib_has_str.split('; ')
vitya_want_arr = vitya_want_str.split('; ')

all_books_arr = needed_arr.copy()
all_books_arr.extend(lib_has_arr)
all_books_arr.extend(vitya_want_arr)
all_books_arr = list(set(all_books_arr))

res_arr = []

for book in all_books_arr:
    if book in needed_arr and book in lib_has_arr and book not in vitya_want_arr:
        res_arr.append(book)
    
res_arr.sort(reverse=True)
for i in res_arr:
    print(i)
```
