```python
import fnmatch

with open("./data/file1.txt", "r") as f:
    data = f.read()
    
arr = data.split('\n')
result = []

#template = input('Напишите строку-шаблон для поиска: ')
template = 'hel*'

for str in arr:
    if fnmatch.fnmatch(str, template):
        result.append(str)

with open("./data/file2.txt", "w+") as f:
    f.write('\n'.join(result))
```
