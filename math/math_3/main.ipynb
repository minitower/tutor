## Критерий согласия Пирсона (χ²)

**Назначение:**
Проверка гипотезы ($H_0$): наблюдаемое распределение соответствует теоретическому.

### Формула

$$
\chi^2 = \sum_{i=1}^{k} \frac{(O_i - E_i)^2}{E_i}
$$

где:

* ($O_i$) — наблюдаемая частота
* ($E_i$) — ожидаемая частота
* ($k$) — число категорий

**Степени свободы:**
$$
\nu = k - 1 \quad (\text{или } k - 1 - m,\ \text{если параметры оценивались})
$$

### Решение

* Вычислить ($\chi^2$)
* Найти *p-value*
* Если (p < $\alpha$) → отвергаем (H_0)

**Применения:**
анализ распределений, проверка моделей, физика частиц, статистический контроль данных.

```python
import numpy as np
from scipy.stats import chisquare

observed = np.array([32, 33, 35])
expected = np.array([30, 35, 35])

chi2, p_value = chisquare(f_obs=observed, f_exp=expected)

print("χ² =", chi2)
print("p-value =", p_value)
```

```python
chi_arr = []
for o, e in zip(observed, expected):
    chi_arr.append((o-e)**2/e)
sum(chi_arr)
```

```python
from scipy.stats import chi2

df = len(observed)-1

p_value = 1-chi2.cdf(sum(chi_arr), df)
print('p-value: ', p_value)
```

## Однофакторный дисперсионный анализ (ANOVA)

**Назначение:**
Проверка гипотезы (H_0): средние значения **нескольких групп равны**.



## Идея метода

ANOVA сравнивает:

* **межгрупповую дисперсию** (различия между средними)
* **внутригрупповую дисперсию** (разброс внутри групп)

### F-статистика

$$
F = \frac{MS_{\text{between}}}{MS_{\text{within}}}
$$

где:
$$
MS_{\text{between}} = \frac{SS_{\text{between}}}{k-1}, \quad
MS_{\text{within}} = \frac{SS_{\text{within}}}{N-k}
$$


## Интерпретация

При ($\alpha$ = 0.05):

$$
p < 0.05 \Rightarrow \text{гипотеза } H_0 \text{ отвергается}
$$

➡ **Средние значимо различаются** (как минимум одна группа отличается).

### Когда использовать ANOVA

* ≥ 3 группы
* Нормальность (примерно)

```python
import numpy as np
from scipy.stats import f_oneway

group1 = np.array([8, 9, 6, 10])
group2 = np.array([4, 5, 6, 5])
group3 = np.array([10, 11, 9, 10])

F_stat, p_value = f_oneway(group1, group2, group3)

print("F-statistic:", F_stat)
print("p-value:", p_value)
```

```python
groups = [group1, group2, group3]

k = len(groups)                         # число групп
n_i = [len(g) for g in groups]          # размеры групп
N = sum(n_i)                            # общее число наблюдений

means = [g.mean() for g in groups]      # средние по группам
grand_mean = np.concatenate(groups).mean()  # общее среднее

print("Means:", means)
print("Grand mean:", grand_mean)
```

```python
SS_between = sum(
    n_i[i] * (means[i] - grand_mean) ** 2
    for i in range(k)
)

SS_within = sum(
    ((g - g.mean()) ** 2).sum()
    for g in groups
)

df_between = k - 1
df_within = N - k
```

```python
MS_between = SS_between / df_between
MS_within = SS_within / df_within
```

```python
F = MS_between / MS_within
print("F-statistic:", F)
```

```python
from scipy.stats import f
p_value = 1 - f.cdf(F, df_between, df_within)
print("p_value:", p_value)
```
