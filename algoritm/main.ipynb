```python
count=0
nums=[-1,-1,1]
prefix = []
nums.sort()
idx = 0
k=0

while idx!=len(nums) and len(prefix)<=len(nums):
    for index, num in enumerate(nums[idx:]):
        print(prefix, count, idx)
        
        if index==0:
            prefix.append(num)
        else:
            prefix.append(num+prefix[-1])
            
        if prefix[-1]<k:
            continue
        elif prefix[-1]==k:
            count+=1
            idx+=1
            prefix=[]
            break
        else:
            idx+=1
            prefix=[]
            break
    
count
```
