```python
t_crit_a = t.ppf(1-(0.95/2), df=len(a)-1)
t_crit_b = t.ppf(1-(0.95/2), df=len(b)-1)

s_a = a['VisitPageFlag'].std()
s_b = b['VisitPageFlag'].std()

mean_a = a['VisitPageFlag'].mean()
mean_b = b['VisitPageFlag'].mean()

left_a = mean_a-t_crit_a*(s_a/math.sqrt(len(a)))
right_a = mean_a+t_crit_a*(s_a/math.sqrt(len(a)))

left_b = mean_b-t_crit_b*(s_b/math.sqrt(len(b)))
right_b = mean_b+t_crit_b*(s_b/math.sqrt(len(b)))

interval_a = [left_a, right_a]
interval_b = [left_b, right_b]

print(interval_a)
print(interval_b)
```
