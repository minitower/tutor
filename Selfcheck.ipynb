```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve

def y1(x):
    return 1.2*x**6 - 14*x**5 + 5.2*x**4 - 8.3*x**3 + x + 2

# начальные приближения (важно взять широкий диапазон)
x0_values = np.linspace(-10, 15, 300)

roots = []

for x0 in x0_values:
    root = fsolve(y1, x0)[0]
    root = round(root, 5)  # округляем для устранения дублей
    
    if abs(y1(root)) < 1:  # ВАЖНО!
        root = round(root, 5)
        if round(root, 5) not in roots:
            roots.append(round(root, 5))

roots = np.array(roots)
print("Найденные корни:", roots)

x = np.linspace(-10, 15, 200)
y = y1(x)

plt.figure(figsize=(10, 6))
plt.axhline(0)  # ось Ox
plt.plot(x, y, label='p(x)')
plt.scatter(roots, [0]*len(roots), color='red', zorder=5, label='Корни')
plt.legend()
plt.grid()
plt.xlabel('x')
plt.ylabel('p(x)')
plt.title('График полинома и его корни')
plt.show()
```

```python
import numpy as np

def y1(x):
    return 1.2*x**6 - 14*x**5 +5.2*x**4-8.3*x**3+x+2


def f(x, coeffs):
    # зеркалим порядок следования коэффициентов,
    # т.к. суммируем начиная с 0-й степени x в цикле for
    # и превращаем список coeffs в массив c:
    c = np.array(coeffs[::-1])
    return np.sum([p*(x**i) for i, p in enumerate(c)])

coeffs = [1.2, -14, 5.2, -8.3, 0, 1, 2]

roots = np.roots(coeffs)

x = np.linspace(-10, 15, 200)
y = y1(x)

plt.figure(figsize=(10, 6))
plt.axhline(0)  # ось Ox
plt.plot(x, y, label='p(x)')
plt.scatter(roots, [0]*len(roots), color='red', zorder=5, label='Корни')
plt.legend()
plt.grid()
plt.xlabel('x')
plt.ylabel('p(x)')
plt.title('График полинома и его корни')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve

# параметр
w = 8

# функция
def f(x):
    return w * np.cos(x) + 3 * x**2 * np.sin(x)

# интервал
a, b = -8, 8

# начальные приближения (сеткой)
initial_guesses = np.linspace(a, b, 50)

roots = []

for guess in initial_guesses:
    root = fsolve(f, guess)[0]
    
    # проверяем, что корень в интервале
    if a <= root <= b:
        # убираем дубликаты
        if not any(abs(root - r) < 0.01 for r in roots):
            roots.append(root)

# сортируем
roots = sorted(roots)

print("Найденные корни:")
for r in roots:
    print(round(r, 6))

# --- график ---
x = np.linspace(a, b, 1000)
y = f(x)

plt.figure()
plt.axhline(0)
plt.plot(x, y, label='f(x)')

# точки корней
for r in roots:
    plt.scatter(r, 0)

plt.title("График функции и корни (метод Ньютона / fsolve)")
plt.legend()
plt.grid()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

w = 8

# сетка
x = np.linspace(-1, 12, 400)
y = np.linspace(-1, 12, 400)

X, Y = np.meshgrid(x, y)

# комплексное число
Z = X + 1j * Y

# функция из задания 2
F = y1(Z)

# модуль
F_abs = np.abs(F)

# логарифм (чтобы было видно корни)
F_log = np.log(F_abs + 1e-6)

# график
plt.figure()
plt.contourf(X, Y, F_log, levels=50)

plt.colorbar(label='log(|f(z)|)')
plt.title("Контурный график |f(z)|")
plt.xlabel("Re(z)")
plt.ylabel("Im(z)")

plt.show()
```

```python
import numpy as np
from scipy.optimize import minimize_scalar, fmin
import matplotlib.pyplot as plt

def f(x,w=8):
    return w * np.cos(x)+3*x**2 * np.sin(x)

def fm(x,w=8):
    return -f(x,w) # отриц функция
    

(x_min, x_max) = (-8, 8)

f_min=fmin(fm, 7)


result = minimize_scalar(fm, bounds=[x_min, x_max], method='bounded')
print(result)

x = f_min
y = f(x)
print(x,y)

plt.figure(figsize=(10,6))
x_arr=np.linspace(x_min,x_max,1000)
plt.plot(x_arr, f(x_arr))
plt.scatter(x,y)
plt.xlabel('x')
plt.ylabel('y')
```

```python
def maxArea(heights) -> int:
    max_area = 0
    left = 0
    right=1
    
    if len(heights)==2:
        return min(heights)*1
    
    while left!=len(heights)-2:
        for i in range(len(heights)-right):
            diff = (right-left)
            if diff==0:
                diff=1
                
            current_area = min(heights[left],heights[right])*diff
            if current_area>max_area:
                max_area=current_area
            right+=1
        left+=1
        right=left+1
        
    return max_area
```

```python
consonants = ['b', 'c', 'd', 'f', 'g', 'h', 'j', 'k', 'l', 'm', 'n', 'p', 'q', 'r', 's', 't', 'v', 'w', 'x', 'y', 'z']
vowels = ['a', 'e', 'i', 'o', 'u', 'y']

def syllable_split(word):
  syllables = [] #several strings, it is a list
  for char in word:
    syllable = '' # no square brackets, because it is a string
    if char in vowels:
      syllables.append(syllable)
    else:
      syllable += char # += takes syllable and adds char, syllable = syllable + char
  return syllables

syllable_split('bby')
```
